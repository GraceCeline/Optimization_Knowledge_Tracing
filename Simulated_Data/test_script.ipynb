```python
import json

with open('../elo_variable.json', 'r') as Elo_Data:
    elo_data = json.load(Elo_Data)

scenarios = elo_data["scenarios"]
print(elo_data)
```

```python
import pandas as pd

for scenario in scenarios:
    scenario_id    = scenario["id"]
    filepath = f"scenario_{scenario_id}_train_data.csv"
    df = pd.read_csv(filepath)

    success_rate = df["success"].sum() / len(df)
    print(f"Scenario {scenario_id} df size {len(df)} success rate: {success_rate}")
```
