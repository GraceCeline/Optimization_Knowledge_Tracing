```python
import pandas as pd
import os
import glob

def compile_all_kt_results(scenario_ids, output_path='kt_results_summary.csv'):
    """
    Compile results from all KT models (BKT, PFA, DKT) across all scenarios.
    Reads individual CSV files and combines into one summary.
    """
    all_results = []
    
    for scenario_id in scenario_ids:
        for model in ['BKT', 'PFA', 'DKT']:
            filename = f'{model.lower()}_results_scenario_{scenario_id}.csv'
            
            if os.path.exists(filename):
                print(f"Loading {filename}...")
                df = pd.read_csv(filename)
                df['model'] = model
                all_results.append(df)
            else:
                print(f"⚠ Warning: {filename} not found")
    
    if not all_results:
        print("❌ No results found!")
        return None
    
    # Combine all results
    combined_df = pd.concat(all_results, ignore_index=True)
    
    # Reorder columns: scenario, model, then rest
    cols = ['scenario', 'model'] + [col for col in combined_df.columns if col not in ['scenario', 'model']]
    combined_df = combined_df[cols]
    
    # Sort by scenario and model
    combined_df = combined_df.sort_values(['scenario', 'model']).reset_index(drop=True)
    
    # Save to CSV
    combined_df.to_csv(output_path, index=False)
    print(f"\n✓ CSV saved: {os.path.abspath(output_path)}")
    
    # Save as formatted TXT
    txt_path = output_path.replace('.csv', '.txt')
    with open(txt_path, 'w') as f:
        f.write("="*160 + "\n")
        f.write("KNOWLEDGE TRACING MODELS COMPARISON - OPTIMIZATION RESULTS\n")
        f.write("="*160 + "\n\n")
        f.write(combined_df.to_string(index=False))
        f.write("\n\n" + "="*160 + "\n")
        f.write(f"Summary:\n")
        f.write(f"  Scenarios: {sorted(combined_df['scenario'].unique())}\n")
        f.write(f"  Models: {sorted(combined_df['model'].unique())}\n")
        f.write(f"  Total rows: {len(combined_df)}\n")
    
    print(f"✓ TXT saved: {os.path.abspath(txt_path)}\n")
    
    return combined_df

```

```python

# ============ RUN COMPILATION ============
scenario_ids = [1, 2, 3, 4, 5, 6]  # Your scenario IDs

final_table = compile_all_kt_results(
    scenario_ids=scenario_ids,
    output_path='kt_results_summary.csv'
)

print("\n" + "="*160)
print("FINAL COMBINED RESULTS")
print("="*160)
print(final_table)
```
